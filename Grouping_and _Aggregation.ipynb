# groupby()

## Step 1: Understand the Concept

### What Is `groupby()`?

The `groupby()` method is used to divide a DataFrame into groups based on one or more columns.

After creating groups, aggregation functions such as `sum()`, `mean()`, `count()`, `min()`, and `max()` can be used to analyze each group.

### Why Do We Use It?

- To perform category-wise analysis.
- To calculate totals for different groups.
- To compare groups in a dataset.
- To summarize large datasets.

### Method Used

The `groupby()` method groups rows that have the same value in the selected column.

### Basic Syntax

`df.groupby("Column")["Value_Column"].sum()`

The grouping column defines the groups, and the aggregation function calculates a result for each group.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a shop has sales records from different cities. Instead of checking every record separately, the shop manager puts all Chennai records together, all Madurai records together, and so on. The manager can then calculate the total sales for each city.

### Simple Example

| City | Product | Sales |
|---|---|---:|
| Chennai | Laptop | 50000 |
| Madurai | Phone | 20000 |
| Chennai | Tablet | 30000 |
| Madurai | Laptop | 15000 |

If we group the data by City and calculate the total Sales:

| City | Total Sales |
|---|---:|
| Chennai | 80000 |
| Madurai | 35000 |

### Business Example

A company can group sales records by city, product category, or sales team to understand how much each group contributes to total business sales.

### AI/ML Use Case

`groupby()` is useful during exploratory data analysis and feature engineering. It can help calculate group-level statistics such as average purchase amount for each customer segment.

````python
import pandas as pd

# Create a sample sales DataFrame
sales_data = {
    "City": ["Chennai", "Madurai", "Chennai", "Madurai"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop"],
    "Sales": [50000, 20000, 30000, 15000]
}

sales_df = pd.DataFrame(sales_data)

# Group the data by City and calculate total sales
city_sales = sales_df.groupby("City")["Sales"].sum()

# Display the result
print(city_sales)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(sales_data)` creates a DataFrame containing city, product, and sales information.
3. `groupby("City")` groups the rows based on the City column.
4. `["Sales"]` selects the Sales column for analysis.
5. `sum()` calculates the total sales for each city.
6. The result is stored in `city_sales`.
7. `print(city_sales)` displays the city-wise totals.

### What the Code Does

The code groups the sales records by city and calculates the total sales for each city.

### Why the Method Is Used

The `groupby()` method is used when we need to perform analysis separately for different categories or groups. Combining it with aggregation functions such as `sum()` makes it easy to create summarized business information.

### Expected Output

```text
City
Chennai    80000
Madurai    35000
Name: Sales, dtype: int64
```

Chennai has total sales of `80000`, while Madurai has total sales of `35000`.

### Where It Is Used in AI/ML or Data Analytics

`groupby()` is widely used in exploratory data analysis, business reporting, customer analysis, and feature engineering to calculate group-level statistics.

# agg()

## Step 1: Understand the Concept

### What Is `agg()`?

The `agg()` method stands for aggregation. It is used to apply one or more aggregation functions to grouped or selected data.

### Why Do We Use It?

- To perform multiple calculations at the same time.
- To calculate summary statistics for groups.
- To combine functions such as `sum()`, `mean()`, `min()`, and `max()`.
- To create a detailed summary of grouped data.

### Method Used

The `agg()` method can apply multiple aggregation functions to the same column.

### Basic Syntax

`df.groupby("Group_Column")["Value_Column"].agg(["sum", "mean", "max"])`

This calculates the sum, average, and maximum value for each group.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a store manager wants to compare different city branches. For each city, the manager wants to know the total sales, average sales, and highest single sale. Instead of calculating them separately, all three calculations can be performed together.

### Simple Example

| City | Sales |
|---|---:|
| Chennai | 50000 |
| Madurai | 20000 |
| Chennai | 30000 |
| Madurai | 15000 |

Using `agg()`:

| City | Total Sales | Average Sales | Highest Sale |
|---|---:|---:|---:|
| Chennai | 80000 | 40000 | 50000 |
| Madurai | 35000 | 17500 | 20000 |

### Business Example

A company can use `agg()` to create a city-wise sales report containing total revenue, average transaction value, and highest transaction value.

### AI/ML Use Case

`agg()` can be used during exploratory data analysis and feature engineering to generate multiple group-level statistics that may be useful for understanding patterns in the data.

````python
import pandas as pd

# Create a sample sales DataFrame
sales_data = {
    "City": ["Chennai", "Madurai", "Chennai", "Madurai"],
    "Sales": [50000, 20000, 30000, 15000]
}

sales_df = pd.DataFrame(sales_data)

# Group by City and calculate multiple statistics
city_summary = sales_df.groupby("City")["Sales"].agg(
    ["sum", "mean", "max"]
)

# Display the result
print(city_summary)
````

## Step 3: Implement the Concept — Code Explanation

### Code Explanation

1. `import pandas as pd` imports the Pandas library.
2. `pd.DataFrame(sales_data)` creates the sales DataFrame.
3. `groupby("City")` groups the records based on city.
4. `["Sales"]` selects the Sales column.
5. `agg(["sum", "mean", "max"])` performs three calculations for each city.
6. `sum` calculates total sales.
7. `mean` calculates average sales.
8. `max` finds the highest sales value.
9. The result is stored in `city_summary`.

### What the Code Does

The code creates a city-wise summary containing total sales, average sales, and the highest sale for each city.

### Why the Method Is Used

The `agg()` method is useful when multiple summary calculations are required for the same grouped data. It avoids writing separate statements for each aggregation.

### Expected Output

```text
             sum     mean    max
City
Chennai    80000  40000.0  50000
Madurai    35000  17500.0  20000
```

For Chennai, the total sales are `80000`, the average is `40000`, and the highest sale is `50000`. The same calculations are performed for Madurai.

### Where It Is Used in AI/ML or Data Analytics

`agg()` is useful for creating summary reports, exploratory data analysis, and group-level features during data preprocessing and feature engineering.

# transform()

## Step 1: Understand the Concept

### What is transform()?

The `transform()` method performs a calculation within each group and returns the result aligned with the original DataFrame.

### Why do we use transform()?

We use `transform()` when we want to calculate a group-level value but keep the same number of rows as the original DataFrame.

For example, we can calculate the average sales for each city and add that average to every row belonging to that city.

### Syntax

```python
df.groupby("column")["value_column"].transform("function")





## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has employees from different departments.

If we calculate the average salary for each department and want to show that department average beside every employee, we can use `transform()`.

Each employee keeps their original row, while the department average is added to that row.

### Simple Example

Suppose we have sales from two cities:

| City | Sales |
|------|------:|
| Chennai | 50,000 |
| Madurai | 20,000 |
| Chennai | 30,000 |
| Madurai | 15,000 |

The city-wise average sales are:

- Chennai → 40,000
- Madurai → 17,500

Using `transform("mean")`, these averages are added to every corresponding row.

### Business Example

A company can calculate the average sales for each city and compare every individual sale with the average sales of that city.

This helps identify sales that are higher or lower than the city average.

### AI/ML Use Case

In AI/ML and Data Analytics, `transform()` can be used to create group-based features.

For example, we can calculate the average purchase amount for each customer group and use it as an additional feature for analysis or machine learning.

````python
import pandas as pd

# Create sales data
sales_data = {
    "City": ["Chennai", "Madurai", "Chennai", "Madurai"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop"],
    "Sales": [50000, 20000, 30000, 15000]
}

sales_df = pd.DataFrame(sales_data)

# Calculate the average sales for each city
sales_df["City_Average_Sales"] = sales_df.groupby("City")["Sales"].transform("mean")

print(sales_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing the city, product, and sales values.

Then, we group the data using the `City` column and calculate the average sales for each city using `transform("mean")`.

The result is stored in a new column called `City_Average_Sales`.

### What the Code Does

The code calculates the average sales separately for each city and places the corresponding city average beside every original row.

For Chennai:

- 50,000 and 30,000
- Average = 40,000

For Madurai:

- 20,000 and 15,000
- Average = 17,500

### Why the Function/Method Is Used

- `groupby("City")` groups the rows based on city.
- `["Sales"]` selects the Sales column for calculation.
- `transform("mean")` calculates the group-wise average while keeping the original number of rows.
- `mean` calculates the average value.
- `sales_df["City_Average_Sales"]` creates a new column to store the result.

### Expected Output

```text
       City  Product  Sales  City_Average_Sales
0   Chennai   Laptop  50000             40000.0
1   Madurai    Phone  20000             17500.0
2   Chennai   Tablet  30000             40000.0
3   Madurai   Laptop  15000             17500.0

The Chennai rows both receive 40,000 because 40,000 is the average sales for Chennai.

The Madurai rows both receive 17,500 because 17,500 is the average sales for Madurai.

### Where It Is Used in AI/ML or Data Analytics

transform() is useful for creating group-based features while keeping every original record.

For example, customer-level, city-level, department-level, or product-level averages can be added as new features for analysis and machine learning.

# filter()

## Step 1: Understand the Concept

### What is filter()?

The `filter()` method is used after grouping data to keep or remove complete groups based on a condition.

### Why do we use filter()?

We use `filter()` when we want to select groups that satisfy a specific condition.

For example, we can keep only cities whose total sales are greater than 50,000.

### Syntax

```python
df.groupby("column").filter(lambda group: condition)




## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has several branches.

The company wants to keep only branches whose total monthly sales are above 50,000.

Instead of checking each employee or transaction separately, we check the complete branch group.

If a branch satisfies the condition, all its records are kept.

### Simple Example

Suppose we have:

| City | Sales |
|------|------:|
| Chennai | 50,000 |
| Madurai | 20,000 |
| Chennai | 30,000 |
| Madurai | 15,000 |

Total sales:

- Chennai → 80,000
- Madurai → 35,000

If the condition is `total sales > 50,000`, only the Chennai group is kept.

### Business Example

A company can use `filter()` to identify branches, stores, or departments that achieve a minimum sales target.

For example, only branches with total sales above 50,000 can be selected for further analysis.

### AI/ML Use Case

In AI/ML and Data Analytics, `filter()` can be used to remove groups with insufficient data.

For example, we can keep only customers who have made a minimum number of transactions before performing customer behavior analysis.

````python
import pandas as pd

# Create sales data
sales_data = {
    "City": ["Chennai", "Madurai", "Chennai", "Madurai"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop"],
    "Sales": [50000, 20000, 30000, 15000]
}

sales_df = pd.DataFrame(sales_data)

# Keep only cities whose total sales are greater than 50,000
filtered_df = sales_df.groupby("City").filter(
    lambda group: group["Sales"].sum() > 50000
)

print(filtered_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing city, product, and sales information.

Then, we group the data using the `City` column.

For each city group, we calculate the total sales using `sum()`.

The `filter()` method keeps the complete group only when the total sales are greater than 50,000.

### What the Code Does

The code checks the total sales of each city:

- Chennai → 50,000 + 30,000 = 80,000
- Madurai → 20,000 + 15,000 = 35,000

Only Chennai satisfies the condition of total sales greater than 50,000.

Therefore, all Chennai rows are kept and all Madurai rows are removed.

### Why the Function/Method Is Used

- `groupby("City")` groups the data by city.
- `filter()` checks a condition for each complete group.
- `lambda group:` creates a small function for each group.
- `group["Sales"].sum()` calculates the total sales of the group.
- `> 50000` defines the condition for keeping the group.

### Expected Output

```text
       City  Product  Sales
0   Chennai   Laptop  50000
2   Chennai   Tablet  30000

Only the Chennai group remains because its total sales are 80,000, which is greater than 50,000.

### Where It Is Used in AI/ML or Data Analytics

filter() is useful when selecting meaningful groups for analysis.

For example, analysts can keep customers with enough transactions, stores with sufficient sales, or departments with enough records before performing further analysis or machine learning.

# Pivot Tables

## Step 1: Understand the Concept

### What is a Pivot Table?

A Pivot Table is used to summarize and organize data based on different categories.

It helps us compare values across rows and columns.

### Why do we use Pivot Tables?

We use Pivot Tables to quickly summarize large datasets and understand relationships between different categories.

For example, we can calculate total sales for each product across different cities.

### Function Used

Pandas provides the `pd.pivot_table()` function to create Pivot Tables.

### Syntax

```python
pd.pivot_table(
    data,
    index="row_column",
    columns="column_column",
    values="value_column",
    aggfunc="sum"
)


## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a supermarket manager wants to know how much each product was sold in different branches.

Instead of checking every transaction separately, the manager can create a summary table with branches as rows and products as columns.

This makes comparison easier.

### Simple Example

Suppose we have:

| City | Product | Sales |
|---|---|---:|
| Chennai | Laptop | 50,000 |
| Madurai | Phone | 20,000 |
| Chennai | Tablet | 30,000 |
| Madurai | Laptop | 15,000 |

A Pivot Table can organize this information like:

| City | Laptop | Phone | Tablet |
|---|---:|---:|---:|
| Chennai | 50,000 | 0 | 30,000 |
| Madurai | 15,000 | 20,000 | 0 |

This makes it easier to compare product sales between cities.

### Business Example

A company can use Pivot Tables to analyze sales by city, product, department, or month.

For example, a sales manager can quickly compare which products generate the highest sales in different cities.

### AI/ML Use Case

In AI/ML and Data Analytics, Pivot Tables can be used during Exploratory Data Analysis (EDA) to summarize relationships between categorical variables and numerical values.

For example, product sales can be compared across different regions before building a machine learning model.

````python
import pandas as pd

# Create sales data
sales_data = {
    "City": ["Chennai", "Madurai", "Chennai", "Madurai"],
    "Product": ["Laptop", "Phone", "Tablet", "Laptop"],
    "Sales": [50000, 20000, 30000, 15000]
}

sales_df = pd.DataFrame(sales_data)

# Create a Pivot Table
sales_pivot = pd.pivot_table(
    sales_df,
    index="City",
    columns="Product",
    values="Sales",
    aggfunc="sum",
    fill_value=0
)

print(sales_pivot)
````

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing city, product, and sales information.

Then, we use `pd.pivot_table()` to summarize the sales data.

The `City` column is used as the row category, and the `Product` column is used as the column category.

The `Sales` column is summarized using the `sum` aggregation function.

### What the Code Does

The code creates a table that shows total sales for each product in each city.

### Why the Function/Method Is Used

- `pd.pivot_table()` creates the summary table.
- `index="City"` places cities as rows.
- `columns="Product"` places products as columns.
- `values="Sales"` tells Pandas which numerical values to summarize.
- `aggfunc="sum"` calculates total sales.
- `fill_value=0` replaces missing combinations with 0.

### Expected Output

```text
Product   Laptop  Phone  Tablet
City
Chennai    50000      0   30000
Madurai    15000  20000       0

For example, Chennai has 50,000 sales for Laptop and 30,000 sales for Tablet.

There is no Phone sale recorded for Chennai, so it is shown as 0.

### Where It Is Used in AI/ML or Data Analytics

Pivot Tables are commonly used in Exploratory Data Analysis (EDA) to summarize and compare data.

They can help identify patterns such as product performance across cities, customer behavior across regions, or sales trends across different categories.

# crosstab()

## Step 1: Understand the Concept

### What is crosstab()?

The `pd.crosstab()` function is used to calculate frequency counts between categorical variables.

It creates a table that shows how often categories occur together.

### Why do we use crosstab()?

We use `crosstab()` to understand the relationship between categorical columns by counting their combinations.

For example, we can count how many times each product was purchased in each city.

### Function Used

```python
pd.crosstab()

pd.crosstab(
    index,
    columns
)
### Important Parameters
index → Category shown as rows.
columns → Category shown as columns.

The default result is a frequency count.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a supermarket manager wants to know how many customers purchased each type of product in each city.

A crosstab can create a quick count table showing the number of purchases for each city and product combination.

### Simple Example

Suppose we have:

| City | Product |
|---|---|
| Chennai | Laptop |
| Chennai | Phone |
| Chennai | Laptop |
| Madurai | Laptop |
| Madurai | Phone |

The crosstab will count how many times each product appears in each city.

| City | Laptop | Phone |
|---|---:|---:|
| Chennai | 2 | 1 |
| Madurai | 1 | 1 |

### Business Example

A business can use `crosstab()` to analyze customer choices.

For example, a company can find how many customers from each city purchased different product categories.

### AI/ML Use Case

In AI/ML and Data Analytics, `crosstab()` can be used during Exploratory Data Analysis (EDA) to understand relationships between categorical features.

For example, we can compare customer city with product category or customer segment with purchase behavior.

````python
import pandas as pd

# Create purchase data
purchase_data = {
    "City": ["Chennai", "Chennai", "Chennai", "Madurai", "Madurai"],
    "Product": ["Laptop", "Phone", "Laptop", "Laptop", "Phone"]
}

purchase_df = pd.DataFrame(purchase_data)

# Create a crosstab
purchase_crosstab = pd.crosstab(
    purchase_df["City"],
    purchase_df["Product"]
)

print(purchase_crosstab)
````

## Step 3: Code Explanation

### Code Explanation

First, we create a DataFrame containing city and product information.

Then, we use `pd.crosstab()` to count the number of times each product appears for each city.

### What the Code Does

The code creates a frequency table between `City` and `Product`.

It counts each city-product combination.

### Why the Function/Method Is Used

- `pd.crosstab()` creates a cross-tabulation table.
- `purchase_df["City"]` is used as the row category.
- `purchase_df["Product"]` is used as the column category.
- The function automatically counts the occurrences of each combination.

### Expected Output

```text
Product  Laptop  Phone
City
Chennai        2      1
Madurai        1      1

This means:

Chennai → Laptop appeared 2 times and Phone appeared 1 time.
Madurai → Laptop appeared 1 time and Phone appeared 1 time.

### Where It Is Used in AI/ML or Data Analytics

crosstab() is useful in Exploratory Data Analysis (EDA) for understanding relationships between categorical variables.

It can help analyze customer segments, product preferences, regions, survey responses, and other categorical data before further analysis or machine learning.
