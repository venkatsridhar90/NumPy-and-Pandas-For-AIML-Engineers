# concat()

## Step 1: Understand the Concept

### What is concat()?

The `concat()` function is used to combine two or more Pandas DataFrames or Series into a single DataFrame.

By default, it combines data row-wise.

### Why do we use concat()?

We use `concat()` when we have multiple datasets with a similar structure and want to combine them into one dataset.

For example, January and February sales data can be combined into one sales DataFrame.

### Syntax

```python
pd.concat([df1, df2])



### Important Parameter

axis=0 combines data row-wise.

axis=1 combines data column-wise.

By default, axis=0 is used.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company maintains separate sales records for January and February.

Both files contain the same columns, such as Product and Sales.

Instead of keeping them separately, we can place the February records below the January records and create one combined dataset.

### Simple Example

January:

| Product | Sales |
|---|---:|
| Laptop | 50,000 |
| Phone | 20,000 |

February:

| Product | Sales |
|---|---:|
| Tablet | 30,000 |
| Headphones | 10,000 |

After using `concat()`, the records are combined into one DataFrame.

### Business Example

A company may receive monthly sales files from different branches or departments.

If the files have the same structure, `concat()` can combine them into a single dataset for reporting and analysis.

### AI/ML Use Case

In AI/ML and Data Analytics, `concat()` can be used to combine datasets collected at different times or from different sources when they have a compatible structure.

For example, monthly customer records can be combined before performing data analysis or model training.

````python
import pandas as pd

# Create January sales data
january_data = {
    "Product": ["Laptop", "Phone"],
    "Sales": [50000, 20000]
}

january_df = pd.DataFrame(january_data)

# Create February sales data
february_data = {
    "Product": ["Tablet", "Headphones"],
    "Sales": [30000, 10000]
}

february_df = pd.DataFrame(february_data)

# Combine the DataFrames row-wise
combined_df = pd.concat([january_df, february_df], ignore_index=True)

print(combined_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create two DataFrames: one for January sales and another for February sales.

Both DataFrames have the same columns: `Product` and `Sales`.

Then, we use `pd.concat()` to combine both DataFrames row-wise.

### What the Code Does

The January records are placed first, and the February records are added below them.

The result is one DataFrame containing all four sales records.

### Why the Function/Method Is Used

- `pd.DataFrame()` creates the DataFrames.
- `pd.concat()` combines multiple DataFrames.
- `[january_df, february_df]` provides the DataFrames that need to be combined.
- `ignore_index=True` creates a new continuous index from 0.

### Expected Output

```text
      Product  Sales
0      Laptop  50000
1       Phone  20000
2      Tablet  30000
3  Headphones  10000

The February records are added below the January records, and the index is reset to 0, 1, 2, and 3.

### Where It Is Used in AI/ML or Data Analytics

concat() is useful for combining datasets with similar structures.

It can be used to combine monthly, yearly, branch-wise, or batch-wise data before performing data analysis, visualization, or machine learning.

# merge()

## Step 1: Understand the Concept

### What is merge()?

The `merge()` function is used to combine two DataFrames based on one or more common columns or keys.

It is similar to joining related tables using a common column.

### Why do we use merge()?

We use `merge()` when information about the same records is stored in different DataFrames and we want to combine the related information.

For example, customer details may be stored in one DataFrame and purchase details in another DataFrame.

### Syntax

```python
pd.merge(df1, df2, on="common_column")



### Important Parameter
on → Specifies the common column used to match the two DataFrames.

The common column should contain related values in both DataFrames.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has two registers.

The first register contains customer names and customer IDs.

The second register contains customer IDs and their purchase amounts.

The customer ID is available in both registers.

We can use the customer ID to match the records and create one complete dataset.

### Simple Example

Customer Data:

| Customer_ID | Name |
|---|---|
| C101 | Arun |
| C102 | Priya |
| C103 | Ravi |

Purchase Data:

| Customer_ID | Purchase |
|---|---:|
| C101 | 1500 |
| C102 | 2200 |
| C103 | 1800 |

After merging using `Customer_ID`:

| Customer_ID | Name | Purchase |
|---|---|---:|
| C101 | Arun | 1500 |
| C102 | Priya | 2200 |
| C103 | Ravi | 1800 |

### Business Example

A company may store customer information and purchase transactions in separate datasets.

Using `merge()`, the company can combine customer details with their purchase information using a common Customer ID.

### AI/ML Use Case

In AI/ML and Data Analytics, `merge()` is commonly used during data preparation.

For example, customer demographic data can be merged with transaction data to create a complete dataset for analysis or machine learning.

````python
import pandas as pd

# Create customer data
customer_data = {
    "Customer_ID": ["C101", "C102", "C103"],
    "Name": ["Arun", "Priya", "Ravi"]
}

customer_df = pd.DataFrame(customer_data)

# Create purchase data
purchase_data = {
    "Customer_ID": ["C101", "C102", "C103"],
    "Purchase": [1500, 2200, 1800]
}

purchase_df = pd.DataFrame(purchase_data)

# Merge the DataFrames using Customer_ID
merged_df = pd.merge(
    customer_df,
    purchase_df,
    on="Customer_ID"
)

print(merged_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create two DataFrames.

The first DataFrame contains customer IDs and names.

The second DataFrame contains customer IDs and purchase amounts.

Both DataFrames have `Customer_ID` as a common column.

We use `pd.merge()` to combine the related records using this common column.

### What the Code Does

The code matches the same `Customer_ID` from both DataFrames and combines their information into one DataFrame.

For example, `C101` is matched with `C101`, so Arun's purchase amount is added to the customer record.

### Why the Function/Method Is Used

- `pd.merge()` combines related DataFrames.
- `on="Customer_ID"` tells Pandas which common column should be used for matching.
- The matching Customer IDs are combined into one row.

### Expected Output

```text
  Customer_ID   Name  Purchase
0        C101   Arun      1500
1        C102  Priya      2200
2        C103   Ravi      1800

Each customer record contains both the customer name and the corresponding purchase amount.

### Where It Is Used in AI/ML or Data Analytics

merge() is widely used during data preparation to combine related datasets.

For example, customer information, transaction data, product information, and demographic data can be merged using common IDs before performing analysis or machine learning.

# join()

## Step 1: Understand the Concept

### What is join()?

The `join()` method is used to combine two DataFrames based mainly on their index.

It allows us to add columns from one DataFrame to another DataFrame.

### Why do we use join()?

We use `join()` when two DataFrames have related data and their indexes can be used to align the records.

It is especially convenient when the related key is already set as the DataFrame index.

### Syntax

```python
df1.join(df2)


### Important Point

By default, join() performs a left join.

The rows from the left DataFrame are kept, and matching data from the right DataFrame is added based on the index.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has two registers.

The first register contains customer names.

The second register contains customer purchase amounts.

Both registers use the same Customer ID as their index.

We can use `join()` to bring the purchase information beside the customer information.

### Simple Example

Customer Data:

| Customer_ID | Name |
|---|---|
| C101 | Arun |
| C102 | Priya |
| C103 | Ravi |

Purchase Data:

| Customer_ID | Purchase |
|---|---:|
| C101 | 1500 |
| C102 | 2200 |
| C103 | 1800 |

Because Customer ID is the index in both DataFrames, `join()` can align the matching indexes and combine the columns.

### Business Example

A company may maintain customer information and customer purchase information in separate DataFrames.

If both DataFrames use the same customer ID as their index, `join()` can combine them easily.

### AI/ML Use Case

In AI/ML and Data Analytics, `join()` can be useful when different feature datasets are already organized using the same index.

For example, customer demographic features can be joined with customer transaction features before analysis or model training.

````python
import pandas as pd

# Create customer data
customer_data = {
    "Customer_ID": ["C101", "C102", "C103"],
    "Name": ["Arun", "Priya", "Ravi"]
}

customer_df = pd.DataFrame(customer_data)

# Set Customer_ID as the index
customer_df = customer_df.set_index("Customer_ID")

# Create purchase data
purchase_data = {
    "Customer_ID": ["C101", "C102", "C103"],
    "Purchase": [1500, 2200, 1800]
}

purchase_df = pd.DataFrame(purchase_data)

# Set Customer_ID as the index
purchase_df = purchase_df.set_index("Customer_ID")

# Join the DataFrames using their index
joined_df = customer_df.join(purchase_df)

print(joined_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create two DataFrames.

The first DataFrame contains customer names.

The second DataFrame contains purchase amounts.

We use `set_index()` to make `Customer_ID` the index of both DataFrames.

Then, `join()` combines the DataFrames by matching their indexes.

### What the Code Does

The code matches the same `Customer_ID` indexes and adds the purchase information to the customer DataFrame.

For example, the index `C101` in the first DataFrame matches `C101` in the second DataFrame.

### Why the Function/Method Is Used

- `set_index("Customer_ID")` makes Customer ID the DataFrame index.
- `join()` combines the DataFrames using their indexes.
- The default left join keeps all rows from the left DataFrame.

### Expected Output

```text
              Name  Purchase
Customer_ID
C101          Arun      1500
C102         Priya      2200
C103          Ravi      1800

The purchase information is added beside each customer's name by matching the Customer ID index.

### Where It Is Used in AI/ML or Data Analytics

join() is useful for combining related feature datasets that share the same index.

It can be used to combine customer features, transaction features, or other prepared datasets before analysis or machine learning.

# append() - Deprecation and Alternatives

## Step 1: Understand the Concept

### What is append()?

`append()` was an older Pandas method used to add the rows of one DataFrame to another DataFrame.

### Why was append() used?

It was previously used to combine DataFrames row-wise.

For example:

```python
df1.append(df2)

would add the rows of df2 after the rows of df1.

### Deprecation

The DataFrame.append() method was deprecated in Pandas 1.4 and removed in Pandas 2.0.

Therefore, append() should not be used in new Pandas code.

### Recommended Alternative

Use pd.concat() instead.

pd.concat([df1, df2])
###Important Point

append() and concat() were commonly used for similar row-wise combining tasks, but pd.concat() is the recommended modern approach.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine you have two lists of sales records.

Earlier, you could think of `append()` as taking the second list and adding it to the end of the first list.

In modern Pandas, `concat()` is the recommended way to perform this operation.

### Simple Example

Suppose we have:

**January Data**

| Product | Sales |
|---|---:|
| Laptop | 50,000 |
| Phone | 20,000 |

**February Data**

| Product | Sales |
|---|---:|
| Tablet | 30,000 |
| Headphones | 10,000 |

The old approach was:

```python
df1.append(df2)

The recommended modern approach is:

pd.concat([df1, df2])
### Business Example

A company may have monthly sales DataFrames that need to be combined into one dataset.

Older projects may use append(), but modern applications should use pd.concat().

### AI/ML Use Case

In AI/ML and Data Analytics, existing projects or older notebooks may contain append().

Understanding its deprecation helps us update old code and replace it with pd.concat() when preparing datasets.

````python
#python id="f3v9xa"
import pandas as pd

# Create January sales data
january_data = {
    "Product": ["Laptop", "Phone"],
    "Sales": [50000, 20000]
}

january_df = pd.DataFrame(january_data)

# Create February sales data
february_data = {
    "Product": ["Tablet", "Headphones"],
    "Sales": [30000, 10000]
}

february_df = pd.DataFrame(february_data)

# Recommended modern alternative to append()
combined_df = pd.concat(
    [january_df, february_df],
    ignore_index=True
)

print(combined_df)
````

## Step 3: Code Explanation

### Code Explanation

We create two DataFrames containing January and February sales data.

The old `append()` method would have been used to add the February rows after the January rows.

However, `append()` is no longer available in modern Pandas.

Therefore, we use `pd.concat()` as the recommended alternative.

### What the Code Does

The code combines the January and February DataFrames row-wise into one DataFrame.

### Why the Function/Method Is Used

- `pd.concat()` combines multiple DataFrames.
- `[january_df, february_df]` specifies the DataFrames to combine.
- `ignore_index=True` creates a new continuous index.
- `append()` is not used because it was removed from Pandas 2.0.

### Expected Output

```text
      Product  Sales
0      Laptop  50000
1       Phone  20000
2      Tablet  30000
3  Headphones  10000

The February records are added below the January records.

### Where It Is Used in AI/ML or Data Analytics

Understanding append() deprecation is useful when working with older Pandas projects.

When updating older data-processing code, append() operations should generally be replaced with pd.concat() to maintain compatibility with modern Pandas versions.

# combine_first()

## Step 1: Understand the Concept

### What is combine_first()?

The `combine_first()` method combines two DataFrames by filling missing values in the first DataFrame with corresponding values from the second DataFrame.

### Why do we use combine_first()?

We use `combine_first()` when we have two datasets containing overlapping information and want to use the second dataset as a backup for missing values in the first dataset.

### Syntax

```python
df1.combine_first(df2)                                                                                                                                                                                                                                                            

### Important Point

The values from the first DataFrame are given priority.

If a value in the first DataFrame is missing, Pandas uses the corresponding value from the second DataFrame.

## Step 2: Demonstrate the Concept

### Real-World Analogy

Imagine a company has a main customer record and a backup customer record.

If the main record already contains a customer's city, we keep it.

If the city is missing in the main record, we check the backup record and use the available city.

### Simple Example

Main Data:

| Customer | City | Age |
|---|---|---:|
| Arun | Chennai | 25 |
| Priya | Missing | 28 |
| Ravi | Madurai | Missing |

Backup Data:

| Customer | City | Age |
|---|---|---:|
| Arun | Chennai | 25 |
| Priya | Madurai | 28 |
| Ravi | Coimbatore | 24 |

After using `combine_first()`:

| Customer | City | Age |
|---|---|---:|
| Arun | Chennai | 25 |
| Priya | Madurai | 28 |
| Ravi | Madurai | 24 |

Only the missing values from the main DataFrame are filled using the backup DataFrame.

### Business Example

A company may have customer information from two different systems.

If some customer details are missing in the primary dataset, information from another dataset can be used to fill those missing values.

### AI/ML Use Case

In AI/ML and Data Analytics, `combine_first()` can help combine complementary datasets and fill missing feature values when a reliable secondary source is available.

````python
import pandas as pd
import numpy as np

# Create the primary customer data
primary_data = {
    "Customer": ["Arun", "Priya", "Ravi"],
    "City": ["Chennai", np.nan, "Madurai"],
    "Age": [25, 28, np.nan]
}

primary_df = pd.DataFrame(primary_data).set_index("Customer")

# Create the backup customer data
backup_data = {
    "Customer": ["Arun", "Priya", "Ravi"],
    "City": ["Chennai", "Madurai", "Coimbatore"],
    "Age": [25, 28, 24]
}

backup_df = pd.DataFrame(backup_data).set_index("Customer")

# Fill missing values in the primary DataFrame
combined_df = primary_df.combine_first(backup_df)

print(combined_df)
````

## Step 3: Code Explanation

### Code Explanation

First, we create a primary DataFrame containing customer information.

Some values in the primary DataFrame are missing.

Then, we create a backup DataFrame containing the same customer information with additional available values.

Finally, `combine_first()` fills the missing values in the primary DataFrame using values from the backup DataFrame.

### What the Code Does

The primary DataFrame has priority.

For example:

- Priya's City is missing in the primary DataFrame, so `Madurai` is taken from the backup DataFrame.
- Ravi's Age is missing in the primary DataFrame, so `24` is taken from the backup DataFrame.
- Ravi's City is already available as `Madurai`, so the original value is kept.

### Why the Function/Method Is Used

- `np.nan` represents missing numerical or data values.
- `set_index("Customer")` makes Customer the index so corresponding records can be aligned.
- `combine_first()` fills missing values using the second DataFrame.
- The first DataFrame always has priority when its value is available.

### Expected Output

```text
             City   Age
Customer
Arun      Chennai  25.0
Priya     Madurai  28.0
Ravi      Madurai  24.0

The missing City value for Priya and missing Age value for Ravi are filled using the backup DataFrame.

Existing values from the primary DataFrame are preserved

### Where It Is Used in AI/ML or Data Analytics

combine_first() can be used to combine complementary datasets and fill missing information.

It is useful when preparing cleaner datasets for analysis, feature engineering, and machine learning.
