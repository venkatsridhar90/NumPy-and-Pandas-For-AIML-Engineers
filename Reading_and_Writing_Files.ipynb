<h1>Reading & Writing Data </h1>


```python
import pandas as pd
import numpy as np
```

```python
students = pd.DataFrame({
    "ID":[1,2,3,4],
    "Name":["Arjun","Rogan","Prakesh","Magesh"],
    "Age":[23,20,24,22],
    "Department": ["Civil","Mechanical","Computer","Electronic"],
    "Marks":[80,90,78,89]
})

print(students)
```

<h2>CSV Files</h2>

CSV (Comma Separated Values) stores tabular data where each value is separated by commas.

It is the most commonly used format in Data Science and Machine Learning.

__Example:__

An e-commerce company exports its daily sales report as a CSV file.

<h3>Write CSV</h3>

```python
students.to_csv("students.csv", index=False)

print("CSV file created successfully.")
```

<h3>Read CSV</h3>

```python
df = pd.read_csv("students.csv")

print(df)
```

<h3>AI/ML Usage</h3>

Most datasets downloaded from Kaggle are stored as CSV files.

<h2>Excel Files</h2>

Excel files store data in worksheets with the extensions .xlsx or .xls.

<h2>JSON Files</h2>

JSON (JavaScript Object Notation) stores data as key-value pairs.

It is widely used in APIs and web applications.

__Example:__

A weather API returns temperature, humidity, and wind speed in JSON format.

<h3>Write JSON</h3>

```python
students.to_json(
    "students.json",
    orient="records",
    indent=4
)

print("JSON file created.")
```

<h3>Read JSON</h3>


```python
df = pd.read_json("students.json")

print(df)
```

<h3>AI/ML Usage</h3>

<h2>SQL Tables</h2>

SQL databases store data in tables consisting of rows and columns.

Pandas can both read from and write to SQL databases.

__Example:__

Banks store customer transactions in SQL databases.

```python
import sqlite3
```

<h3>Create Database Connection</h3>

```python
connection = sqlite3.connect("students.db")
```

<h3>Write DataFrame to SQL Table</h3>

```python
students.to_sql(
    "student_table",
    connection,
    if_exists="replace",
    index=False
)

print("Table created successfully.")
```

<h3>Read SQL Table</h3>

```python
df = pd.read_sql(
    "SELECT * FROM student_table",
    connection
)

print(df)
```

<h3>Close Connection</h3>

```python
connection.close()
```

<h3>AI/ML Usage</h3>

Machine learning projects often retrieve customer, sales, or medical data directly from SQL databases.

<h2>Reading Multiple Files</h2>

Sometimes data is split across multiple files, and we need to combine them into one DataFrame.

Example:

A supermarket stores one sales CSV file for each month.

January.csv

February.csv

March.csv

The analyst combines them into one dataset.

<h3>Create Two Sample Files</h3>

```python
sales1 = pd.DataFrame({
    "Product": ["Laptop", "Phone"],
    "Sales": [100, 200]
})

sales2 = pd.DataFrame({
    "Product": ["Tablet", "Watch"],
    "Sales": [150, 120]
})

sales1.to_csv("sales1.csv", index=False)
sales2.to_csv("sales2.csv", index=False)

print(sales1)
print()
print(sales2)
```

<h3>Read Multiple Files</h3>

```python
df1 = pd.read_csv("sales1.csv")

df2 = pd.read_csv("sales2.csv")

combined = pd.concat([df1, df2], ignore_index=True)

print(combined)
```
