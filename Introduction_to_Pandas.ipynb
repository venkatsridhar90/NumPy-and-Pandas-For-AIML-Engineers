# Introduction to Pandas

<h2>What is Pandas?</h2>

* Pandas is a Python library used to store, clean, analyze, and manipulate tabular data.

The two most important Pandas objects are:

<ul><li>Series - one-dimensional data (one column)</li>
<li>DataFrame - two-dimensional data (rows + columns)</li></ul>

Pandas simplifies tasks such as:

<ul><li>Reading data from files</li>
<li>Cleaning missing values</li>
<li>Filtering data</li>
<li>Sorting data</li>
<li>Grouping data</li>
<li>Performing statistical analysis</li></ul>

Pandas is built on top of the NumPy library, making it fast and efficient.

### Example

* Imagine a school has a spreadsheet containing information about 1,000 students.

* Instead of manually calculating averages or finding the highest mark, Pandas can perform these operations in just a few lines of code.

```python
import pandas as pd

students = {
    "Name": ["Arjun", "Rogan", "Pranesh"],
    "Marks": [95, 88, 91]
}

df = pd.DataFrame(students)

print(df)
```

<h2>Series</h2>

<h2>Creating DataFrames</h2>

Pandas allows DataFrames to be created from different data sources.

<h3>From Dictionary</h3>

```python
import pandas as pd

data = {
    "Name":["Arjun","Rogan"],
    "Age":[22,23]
}

df = pd.DataFrame(data)

print(df)
```

<h3>From List of Lists</h3>

```python
data = [["Arjun",22],["Rogan",23]]

df = pd.DataFrame(
    data,
    columns=["Name","Age"]
)

print(df)
```

<h3>From NumPy Array</h3>

```python
import numpy as np

arr = np.array([["Arjun",22],["Rogan",23]])

df = pd.DataFrame(
    arr,
    columns=["Name","Age"]
)

print(df)
```

__Example__

A company stores employee details in different formats (Excel, database, or arrays). Pandas can convert all of them into DataFrames for analysis.

<h2>Reading Data</h2>

Reading data means importing data from external files into a Pandas DataFrame so that it can be analyzed and processed.

Pandas supports many file formats such as:

<ul><li>CSV</li>
<li>Excel</li>
<li>JSON</li>
<li>SQL Database</li>
<li>Parquet</li></ul>

<h3>Create a Sample DataFrame</h3>

```python
import pandas as pd

students = pd.DataFrame({
    "ID": [101, 102, 103, 104],
    "Name": ["Arjun", "Rogan", "Pranesh", "Jothi"],
    "Age": [22, 23, 21, 24],
    "Marks": [95, 88, 91, 84]
})

print(students)
```

<h3>Create a CSV File</h3>

```python
students.to_csv("students.csv", index=False)
print("Csv File successfully saved.")
```

<h3>Read CSV File</h3>

```python
df = pd.read_csv("students.csv")
print(df)
```

<h3>Create an Excel File</h3>

```python
students.to_excel("students.xlsx", index=False)

print("Excel file created successfully.")
```

<h3>Read Excel File</h3>

```python
df = pd.read_excel("students.xlsx")

print(df)
```

<h3>Create a JSON File</h3>

```python
students.to_json("students.json", orient="records" ,indent=4)

print("JSON file created successfully.")
```

<h3>Read JSON File</h3>

```python
df = pd.read_json("students.json")

print(df)
```

<h2>Writing Data</h2>

Writing data means saving a DataFrame into a file so it can be shared, stored, or used later.

Pandas supports writing to:

<ul><li>CSV</li>
<li>CExcel</li>
<li>CJSON</li></ul>

<h3>Create a New DataFrame</h3>


```python
import pandas as pd

employees = pd.DataFrame({
    "Employee_ID": [1, 2, 3],
    "Name": ["Jothi", "Saran", "Dinesh"],
    "Department": ["IT", "HR", "Finance"],
    "Salary": [50000, 45000, 60000]
})

print(employees)
```

<h3>Write to CSV</h3>

```python
employees.to_csv("employees.csv", index=False)

print("employees.csv saved successfully.")
```

<h3>Write to Excel</h3>

```python
employees.to_excel("employees.xlsx", index=False)

print("employees.xlsx saved successfully.")
```

<h3>Write to JSON</h3>

```python
employees.to_json("employees.json", orient="records", indent=4)

print("employees.json saved successfully.")
```

<h3>Verify the Saved Files</h3>

```python
csv_data = pd.read_csv("employees.csv")

excel_data = pd.read_excel("employees.xlsx")

json_data = pd.read_json("employees.json")

print(csv_data)

print()

print(excel_data)

print()

print(json_data)
```
