# 1. Vectorization

## Step 1 – Understand the Concept

Vectorization means performing an operation on an entire Pandas Series or NumPy array at once instead of processing each value one by one using a loop.

Vectorized operations are generally faster because Pandas and NumPy use optimized operations internally.

### Why is it needed?

- Improves execution speed.
- Reduces the need for Python loops.
- Useful when working with large datasets.
- Commonly used in Data Analytics and AI/ML preprocessing.

## Step 2 – Demonstrate the Concept

### Simple Example
Instead of adding 10 to every value using a loop, we can directly add 10 to the entire column.

### Real-World Example
A company has thousands of product prices and wants to increase every price by 5%.

### Business Example
A sales team wants to calculate updated prices or total sales for thousands of transactions.

### AI/ML Use Case
Feature values can be scaled or transformed efficiently before training a machine learning model.

### Data Analytics Use Case
Vectorization can be used for calculating totals, percentages, profits, and other values across large DataFrames.

## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Price": [50000, 30000, 20000, 15000]
})

# Vectorized operation
df["Updated_Price"] = df["Price"] * 1.05

print(df)
```

## Code Explanation

- `df["Price"] * 1.05` performs the calculation on the complete column.
- No `for` loop is required.
- Pandas applies the operation efficiently to all rows.
- The result is stored in the `Updated_Price` column.

### Expected Output

The prices will be increased by 5%.

### AI/ML
Vectorization is useful for feature transformations and preprocessing.

### Data Analytics
Vectorization is useful for calculations on large columns such as sales, profit, and percentages.

# 2. apply() vs Loops

## Step 1 – Understand the Concept

Both `apply()` and loops can perform operations on individual values.

A Python loop processes values one by one.

The `apply()` method applies a function to values in a Pandas Series or DataFrame.

For simple mathematical operations, vectorization is generally preferred over both loops and `apply()`.

### Why is it needed?

- Helps understand different ways to process data.
- Makes code easier to compare and optimize.
- Helps choose an efficient approach for large datasets.

## Step 2 – Demonstrate the Concept

### Simple Example
Add 5 marks to every student's score.

### Real-World Example
Calculate a discount for every product.

### Business Example
Convert customer transaction amounts into different categories.

### AI/ML Use Case
Apply a custom transformation to a feature when direct vectorization is not suitable.

### Data Analytics Use Case
Create a derived column based on an existing column.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Marks": [70, 80, 65, 90]
})

# Using apply()
df["Updated_Marks"] = df["Marks"].apply(lambda x: x + 5)

print(df)
```

## Code Explanation

- `apply()` applies the given function to each value in the `Marks` column.
- `lambda x: x + 5` adds 5 to each mark.
- `apply()` is useful when the transformation is custom or not easily expressed using vectorization.

### Expected Output

The `Updated_Marks` column will contain:

70 → 75  
80 → 85  
65 → 70  
90 → 95

### Important Performance Point

For a simple operation, vectorization is generally preferred:

`df["Marks"] + 5`

instead of:

`df["Marks"].apply(lambda x: x + 5)`

### AI/ML
`apply()` can be useful for custom feature transformations.

### Data Analytics
`apply()` can be useful for creating calculated columns.

# 3. Memory Optimization

## Step 1 – Understand the Concept

Memory optimization means reducing the amount of computer memory used by a DataFrame without losing important information.

Pandas may use more memory than necessary when columns have larger data types than required.

For example, if a column contains small integer values, a smaller integer data type can sometimes be used.

### Why is it needed?

- Helps process large datasets.
- Reduces memory usage.
- Can improve performance.
- Helps avoid memory-related errors.

## Step 2 – Demonstrate the Concept

### Simple Example
A column contains ages between 18 and 60. Using an unnecessarily large data type can waste memory.

### Real-World Example
A company stores millions of customer records.

### Business Example
A retail company processes a large transaction dataset containing millions of rows.

### AI/ML Use Case
Large training datasets may require memory optimization before model training.

### Data Analytics Use Case
Memory optimization allows analysts to work with larger datasets more efficiently.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 24, 25, 30, 28],
    "Sales": [1000, 2000, 1500, 3000, 2500]
})

print("Memory usage before optimization:")
print(df.memory_usage(deep=True))

# Optimize data types
df["Age"] = df["Age"].astype("int8")
df["Sales"] = df["Sales"].astype("int32")

print("\nMemory usage after optimization:")
print(df.memory_usage(deep=True))
```

## Code Explanation

- `memory_usage(deep=True)` checks the memory used by each column.
- `astype("int8")` changes the `Age` column to a smaller integer type.
- `astype("int32")` changes the `Sales` column to a smaller integer type.
- Smaller data types can reduce memory usage when the values fit within their range.

### Expected Output

The memory usage of the optimized columns should be lower than before optimization.

### AI/ML
Memory optimization is useful when working with large training datasets.

### Data Analytics
It helps analysts process large DataFrames using less memory.

# 4. Category Data Type

## Step 1 – Understand the Concept

The `category` data type is useful for columns containing a limited number of repeated values.

For example:

Chennai  
Mumbai  
Chennai  
Delhi  
Mumbai

Instead of storing every repeated value independently, Pandas can represent the values using categorical data.

### Why is it needed?

- Can reduce memory usage for repeated categorical values.
- Can improve some categorical operations.
- Useful for large datasets.

## Step 2 – Demonstrate the Concept

### Simple Example
A City column contains repeated city names.

### Real-World Example
Customer data may contain thousands of records but only a limited number of cities.

### Business Example
An e-commerce dataset may contain repeated product categories such as Electronics, Clothing, and Grocery.

### AI/ML Use Case
Categorical features can be stored efficiently before preprocessing and encoding.

### Data Analytics Use Case
Useful for repeated values such as city, department, product type, or customer segment.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Customer": ["Arun", "Priya", "Kumar", "Anitha", "Rahul"],
    "City": ["Chennai", "Mumbai", "Chennai", "Delhi", "Mumbai"]
})

print("Before conversion:")
print(df.dtypes)

df["City"] = df["City"].astype("category")

print("\nAfter conversion:")
print(df.dtypes)

print("\nMemory usage:")
print(df.memory_usage(deep=True))
```

## Code Explanation

- Initially, the `City` column is stored as a string/object column.
- `astype("category")` converts the column to the categorical data type.
- Repeated categorical values can be stored more efficiently.
- `memory_usage(deep=True)` helps inspect memory usage.

### Expected Output

The `City` column should show:

`category`

instead of:

`object`

### AI/ML
Categorical data can be prepared for encoding before machine learning.

### Data Analytics
Category data type is useful for repeated values such as city, department, product category, and status.

# 5. Efficient Pandas Practices

## Step 1 – Understand the Concept

Efficient Pandas practices are techniques used to write faster, cleaner, and more memory-efficient Pandas code.

Some important practices are:

- Prefer vectorization over loops.
- Select only required columns.
- Use appropriate data types.
- Use `category` for suitable repeated categorical values.
- Avoid unnecessary copies of large DataFrames.
- Use efficient filtering methods.
- Read large files in chunks when necessary.

### Why is it needed?

Efficient practices are important when working with large datasets because inefficient code can consume more memory and take more time.

## Step 2 – Demonstrate the Concept

### Simple Example
Instead of loading or processing all columns, select only the columns required for analysis.

### Real-World Example
A company has a CSV file containing 50 columns but only needs Product, Sales, and Quantity.

### Business Example
A sales analyst only needs selected columns to calculate revenue and average price.

### AI/ML Use Case
Selecting only relevant features can reduce unnecessary data processing before model training.

### Data Analytics Use Case
Efficient column selection, filtering, and vectorized calculations can improve analysis performance.


## Step 3: Implement the Concept

```python
import pandas as pd

df = pd.DataFrame({
    "Product": ["Laptop", "Phone", "Tablet", "Monitor"],
    "Sales": [50000, 30000, 20000, 15000],
    "Quantity": [5, 10, 7, 6],
    "City": ["Chennai", "Mumbai", "Delhi", "Chennai"]
})

# Select only required columns
sales_data = df[["Product", "Sales", "Quantity"]]

# Vectorized calculation
sales_data["Average_Price"] = (
    sales_data["Sales"] / sales_data["Quantity"]
)

# Filter required rows
high_sales = sales_data[sales_data["Sales"] > 20000]

print("Selected Data:")
print(sales_data)

print("\nHigh Sales:")
print(high_sales)
```
