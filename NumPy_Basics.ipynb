# Introduction to NumPy

## 1. What is NumPy?
* NumPy is a Python library that provides a fast multi-dimensional array object  and a large collection of mathematical functions to work on it.

#### Real-world example:
* A teacher stores the marks of 1,000 students.
* With a plain Python list she needs a loop to add 5 grace marks to everyone.
* With NumPy, one line does it for all students.

#### AI/ML Usage
* Pandas, scikit-learn, TensorFlow, PyTorch, OpenCV and Matplotlib are all built on top of  NumPy arrays.
* Datasets, images, weights and embeddings are all stored as arrays.

```python
import numpy as np

print("NumPy version:", np.__version__)

marks = np.array([45, 62, 78, 90, 55])
print("Original marks :", marks)
print("After +5 grace :", marks + 5)
```

## 2. Why NumPy?
* NumPy is preferred over Python lists because it is
* #### Fast

 operations run in compiled C code
* #### Memory efficient 

 stores items of one type in a continuous block of memory
* #### Convenient 

 whole-array maths without writing loops (vectorization)

### Real-world example:
* An e-commerce site applies a 10% discount to 1 million product prices. A Python loop is slow, a NumPy array does it almost instantly.

### AIML Usage:
* Training a model means doing millions of multiplications and additions.
* Vectorized NumPy code is often 10–100x faster than loops, which makes data preprocessing practical.

```python
import numpy as np

python_list = [1,2,3,4]

numpy_array = np.array([1,2,3,4])

print(type(python_list))
print(type(numpy_array))
```

# 3. NumPy Arrays (ndarray)
*  An ndarray is NumPy's main array object, designed to hold elements in a structured array.
### Real-world example: 
* Store student marks across multiple subjects.
### AI/ML usage: 
* Datasets, model features, and labels are often stored as arrays.

```python
sales = np.array([[120, 150, 170],
                  [ 90, 100, 130]])

print(sales)
print("Type of object :", type(sales))
print("Data type      :", sales.dtype)
```

# 4. Creating Arrays
*  Arrays can be created from lists or generated with functions such as zeros, ones, and arange.
### Real-world example: 
* Create a template for sensor readings.
### AI/ML usage: 
* Used to prepare feature arrays and initialize numerical values.

```python
import numpy as np
print(np.array([1, 2, 3]))
print(np.zeros(3))
print(np.ones(3))
print(np.arange(1, 6))
```

## 5. Dimensions 
* The ndim attribute gives the number of dimensions or axes in an array.
#### Real-world example: 
* Daily sales can be 1D; a customer-feature table can be 2D.
#### AI/ML usage:
* Models often expect samples and features in particular dimensions.

```python
import numpy as np
a = np.array([10, 20, 30])
b = np.array([[10, 20], [30, 40]])
print(a.ndim)
print(b.ndim)
```

## 6. Shape
* The shape attribute gives the number of elements along each dimension.
### Real-world example:
*  A table with 100 rows and 4 columns has shape (100, 4).
### AI/ML usage:
*  Checks that sample and feature counts match model requirements.

```python
register = np.random.randint(35, 100, size=(40, 5))

print("Shape of register :", register.shape)
print("Rows (students)   :", register.shape[0])
print("Columns (subjects):", register.shape[1])
```

## 7. Size
* The size attribute returns the total number of elements in an array.
### Real-world example:
*  Count the measurements stored in a dataset.
### AI/ML usage:
*  Helps check the amount of data being processed.


```python
import numpy as np
data = np.array([[1, 2, 3], [4, 5, 6]])
print(data.size)
```

# 8. Data Types
*  The dtype attribute shows the type used to store array elements.
### Real-world example:
*  Customer counts may be integers while prices use decimals.
### AI/ML usage:
* Correct numerical data types help calculations and model training

```python
import numpy as np
counts = np.array([10, 20, 30])
prices = np.array([10.5, 20.75, 30.0])
print(counts.dtype)
print(prices.dtype)
```

# 9. Indexing

* Indexing selects an element by its position; NumPy positions start at 0.
### Real-world example:
*  Retrieve sales for a particular day.
### AI/ML usage:
*  Access individual samples or feature values.

```python
import numpy as np
sales = np.array([120, 150, 135, 180])
print(sales[0])
print(sales[2])
```

# 10. Slicing 
* Slicing selects a range using start:stop; the stop position is excluded.
### Real-world example:
*  Select sales from the first three days.
### AI/ML usage:
* Select subsets of rows or columns for analysis.

```python
import numpy as np
sales = np.array([120, 150, 135, 180, 200])
print(sales[1:4])
print(sales[:3])
```

# 11. Reshaping
* reshape() changes an array's shape when the number of elements fits the new shape.
### Real-world example: 
* Arrange 12 measurements into 3 rows of 4 values.
### AI/ML usage: 
* Convert data to the shape a model expects.

```python
import numpy as np
data = np.arange(1, 7)
print(data.reshape(2, 3))
```

# 12. Flatten
* flatten() converts a multidimensional array into a 1D copy.
### Real-world example:
* Turn a sales table into a single list.
### AI/ML usage:
* Can convert a matrix or image array into one vector.

```python
import numpy as np
matrix = np.array([[1, 2, 3], [4, 5, 6]])
print(matrix.flatten())
```

# 13. Copy vs View
* A copy has separate data; a view shares data with the original array.
### Real-world example: 
* A copy is a separate spreadsheet; a view refers to the same underlying data.
### AI/ML usage: 
* Understanding this avoids accidental source-data changes during preprocessing.

```python
import numpy as np
original = np.array([10, 20, 30])
copied = original.copy()
viewed = original.view()
copied[0] = 99
viewed[1] = 88
print("Original:", original)
print("Copy:", copied)
print("View:", viewed)
```

# 14. Broadcasting
*  Broadcasting allows operations on arrays with compatible shapes without manually repeating values.
### Real-world example:
* Apply the same tax rate to every product price.
### AI/ML usage:
* Used for scaling features and applying offsets efficiently.

```python
import numpy as np
prices = np.array([100, 200, 300])
print(prices * 1.10)
```

# 15. Array Operations
*  Arithmetic and comparison operations can be applied element by element to arrays.
### Real-world example:
*  Calculate profit for multiple products at once.
### AI/ML usage:
* Used in preprocessing and calculations on features and predictions.

```python
import numpy as np
sales = np.array([100, 200, 300])
costs = np.array([60, 120, 180])
print("Profit:", sales - costs)
print("Double sales:", sales * 2)
```

# 16. Universal Functions (ufuncs)
* Universal functions perform element-wise mathematical operations on arrays.
### Real-world example:
* Calculate square roots or round many values at once.
### AI/ML usage:
* Useful for fast mathematical transformations in numerical workflows.

```python
import numpy as np
values = np.array([1, 4, 9, 16])
print(np.sqrt(values))
print(np.mean(values))
print(np.round(np.array([2.345, 4.678]), 2))
```
