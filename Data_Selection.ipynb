# Topic 1 - loc

## Step 1 – Understand the Concept

loc is used to select rows and columns from a DataFrame using their labels.

The general syntax is:

df.loc[row_label, column_label]

We can use `loc to:
- Select a specific row
- Select specific columns
- Select multiple rows
- Select rows and columns using labels

loc is label-based selection.


## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of an employee database where each employee has an Employee ID such as E001, E002, and E003. If we want the details of employee E001, we can use the Employee ID to find that record.

### Simple Example

Suppose we have the following customer table:

| Customer ID | Name | Age | Purchase |
|-------------|------|-----|----------|
| C001 | Arun | 25 | 1500 |
| C002 | Priya | 28 | 2200 |
| C003 | Ravi | 24 | 1800 |

If we want the details of customer C002, we can use its label with loc.

### Business Example

A company may use customer IDs or employee IDs as labels. `loc can be used to select a specific customer's or employee's information from a DataFrame.

### AI/ML Use Case

In AI/ML and Data Analytics, loc is useful for selecting specific rows or columns by their labels during data preparation, analysis, and feature selection.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800]
}

# Create a DataFrame with custom index labels
customer_df = pd.DataFrame(
    customer_data,
    index=["C001", "C002", "C003"]
)

# Select one row using its label
print(customer_df.loc["C002"])

# Select one value using row and column labels
print("Age:", customer_df.loc["C002", "Age"])
```

# Topic -2 iloc

## Step 1 – Understand the Concept

`iloc` is used to select rows and columns from a DataFrame using their integer positions.

The general syntax is:

`df.iloc[row_position, column_position]`

Positions start from `0`.

For example:
- `df.iloc[0]` selects the first row.
- `df.iloc[1]` selects the second row.
- `df.iloc[1, 2]` selects the value at the second row and third column.

`iloc` is position-based selection.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of seats in a classroom. If students are sitting in positions 0, 1, and 2, we can select a student based on their position instead of their name.

### Simple Example

Suppose we have the following customer table:

| Position | Name | Age | Purchase |
|----------|------|-----|----------|
| 0 | Arun | 25 | 1500 |
| 1 | Priya | 28 | 2200 |
| 2 | Ravi | 24 | 1800 |

If we use `iloc[1]`, we select the second row because its position is `1`.

### Business Example

A business may need to select specific rows or columns based on their position while processing a DataFrame. `iloc` is useful when the exact position of the required data is known.

### AI/ML Use Case

In AI/ML, `iloc` can be used to select rows or columns by position while preparing datasets, separating features, and inspecting specific portions of data.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi"],
    "Age": [25, 28, 24],
    "Purchase": [1500, 2200, 1800]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Select the second row using its position
print(customer_df.iloc[1])

# Select the Age value from the second row
print("Age:", customer_df.iloc[1, 1])
```

# Topic 3 - Boolean Indexing

## Step 1 – Understand the Concept

Boolean Indexing is a technique used to select rows from a DataFrame based on a condition.

The condition returns either:

- `True` – the row satisfies the condition
- `False` – the row does not satisfy the condition

For example:

`df["Age"] > 25`

checks whether each value in the Age column is greater than 25.

We can use the condition inside the DataFrame:

`df[df["Age"] > 25]`

This returns only the rows where the condition is True.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a teacher wants to identify students who scored more than 70 marks. The teacher checks each student's mark. Students who satisfy the condition are selected, while the others are not.

### Simple Example

Suppose we have the following customer table:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 20 | 1000 |
| Priya | 25 | 2000 |
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

If the condition is `Age > 25`, only Ravi and Divya satisfy the condition.

| Name | Age | Purchase |
|------|-----|----------|
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

### Business Example

A company may want to identify customers whose purchase amount is greater than ₹2,000. Boolean Indexing can select only those customers for further analysis or marketing.

### AI/ML Use Case

In AI/ML, Boolean Indexing is used to filter datasets based on conditions, such as selecting customers above a certain age, records with high values, or observations that meet specific data-quality conditions.



```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Age": [20, 25, 30, 35],
    "Purchase": [1000, 2000, 3000, 4000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Create a Boolean condition
age_condition = customer_df["Age"] > 25

# Display the Boolean result
print("Boolean condition:")
print(age_condition)

# Filter rows using the condition
filtered_df = customer_df[age_condition]

# Display the filtered DataFrame
print("\nFiltered DataFrame:")
print(filtered_df)
```

# Topic - 4 Filtering Rows

## Step 1 – Understand the Concept

Filtering Rows means selecting only the rows that satisfy a specific condition.

For example:

`df[df["Purchase"] > 2000]`

This selects only the rows where the Purchase value is greater than 2000.

Filtering can be done using conditions such as:

- Greater than `>`
- Less than `<`
- Equal to `==`
- Greater than or equal to `>=`
- Less than or equal to `<=`
- Not equal to `!=`

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a shop owner wants to find customers who spent more than ₹2,000. Instead of checking every customer manually, the owner can apply a condition and select only those customers.

### Simple Example

Suppose we have the following customer table:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 20 | 1000 |
| Priya | 25 | 2000 |
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

If we filter customers whose Purchase is greater than 2000, the result will be:

| Name | Age | Purchase |
|------|-----|----------|
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

### Business Example

A company can filter customers who made high-value purchases, employees with a particular salary range, or products with low sales.

### AI/ML Use Case

In AI/ML, row filtering is used to select relevant records before analysis and model training. For example, we can filter transactions above a certain amount or select records belonging to a particular category.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Age": [20, 25, 30, 35],
    "Purchase": [1000, 2000, 3000, 4000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Filter rows where Purchase is greater than 2000
filtered_df = customer_df[customer_df["Purchase"] > 2000]

# Display the filtered rows
print(filtered_df)
```

# Topic - 5 query()

## Step 1 – Understand the Concept

The `query()` method is used to filter rows in a DataFrame by writing a condition as a string.

For example:

`df.query("Age > 25")`

This returns only the rows where the Age is greater than 25.

`query()` can also be used with multiple conditions.

For example:

`df.query("Age > 25 and Purchase > 2000")`

This returns rows where both conditions are satisfied.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a store manager asking: "Show me customers whose age is above 25." This request is like giving a query to the customer database to return only matching records.

### Simple Example

Suppose we have the following customer table:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 20 | 1000 |
| Priya | 25 | 2000 |
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

If we use the condition `Age > 25`, the result will be:

| Name | Age | Purchase |
|------|-----|----------|
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

### Business Example

A company can use `query()` to find customers with high purchase amounts, employees above a certain age, or products with sales above a specific value.

### AI/ML Use Case

In AI/ML, `query()` can be used to filter relevant records before data preprocessing and model training. It is especially useful when filtering based on multiple conditions.



```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Age": [20, 25, 30, 35],
    "Purchase": [1000, 2000, 3000, 4000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Filter rows using query()
filtered_df = customer_df.query("Age > 25")

# Display the filtered rows
print(filtered_df)
```

### Where It Is Used in AI/ML or Data Analytics

In AI/ML and Data Analytics, query() is used to filter relevant records during exploratory data analysis, data cleaning, preprocessing, and dataset preparation.

# Topic 6 -isin()

## Step 1 – Understand the Concept

The `isin()` method checks whether values in a DataFrame column are present in a given list of values.

For example:

`df["City"].isin(["Chennai", "Madurai"])`

checks whether each city is either Chennai or Madurai.

We can use the result to filter the DataFrame:

`df[df["City"].isin(["Chennai", "Madurai"])]`

This returns only the rows that match the given values.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a company wants to contact customers from Chennai, Madurai, and Coimbatore. Instead of checking each customer one by one, we can provide the list of required cities and select matching customers.

### Simple Example

Suppose we have the following customer table:

| Name | City | Purchase |
|------|------|----------|
| Arun | Chennai | 1500 |
| Priya | Madurai | 2200 |
| Ravi | Coimbatore | 1800 |
| Divya | Chennai | 3000 |
| Kumar | Salem | 2500 |

If we want customers from Chennai and Madurai, the filtered result will be:

| Name | City | Purchase |
|------|------|----------|
| Arun | Chennai | 1500 |
| Priya | Madurai | 2200 |
| Divya | Chennai | 3000 |

### Business Example

A company may want to analyze customers from selected cities, products from selected categories, or employees from selected departments. `isin()` makes this type of filtering simple.

### AI/ML Use Case

In AI/ML, `isin()` can be used to select records belonging to specific categories or groups before analysis and preprocessing.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar"],
    "City": ["Chennai", "Madurai", "Coimbatore", "Chennai", "Salem"],
    "Purchase": [1500, 2200, 1800, 3000, 2500]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Define the cities to select
selected_cities = ["Chennai", "Madurai"]

# Filter rows using isin()
filtered_df = customer_df[
    customer_df["City"].isin(selected_cities)
]

# Display the filtered rows
print(filtered_df)
```

### Where It Is Used in AI/ML or Data Analytics

In AI/ML and Data Analytics, isin() is used to select specific categories or groups from a dataset during data filtering, exploratory analysis, and preprocessing.

# Topic - 7 between()

## Step 1 – Understand the Concept

The `between()` method checks whether values in a column fall within a specified range.

The general syntax is:

`df["Column"].between(start, end)`

By default, both the start and end values are included.

For example:

`df["Age"].between(25, 35)`

selects values from 25 through 35, including 25 and 35.

The result can be used to filter DataFrame rows.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Imagine a company wants to identify employees whose age is between 25 and 35 years. Instead of checking every age separately, we can specify the minimum and maximum age and select the matching employees.

### Simple Example

Suppose we have the following customer table:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 20 | 1000 |
| Priya | 25 | 2000 |
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |
| Kumar | 40 | 5000 |

If we use `between(25, 35)`, the result will be:

| Name | Age | Purchase |
|------|-----|----------|
| Priya | 25 | 2000 |
| Ravi | 30 | 3000 |
| Divya | 35 | 4000 |

### Business Example

A company can use `between()` to find customers whose purchase amount falls within a specific range, such as customers who spent between ₹2,000 and ₹4,000.

### AI/ML Use Case

In AI/ML, `between()` can be used to filter observations within a specific numerical range during data exploration, data cleaning, and preprocessing.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya", "Kumar"],
    "Age": [20, 25, 30, 35, 40],
    "Purchase": [1000, 2000, 3000, 4000, 5000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Filter customers whose age is between 25 and 35
filtered_df = customer_df[
    customer_df["Age"].between(25, 35)
]

# Display the filtered rows
print(filtered_df)
```

### Where It Is Used in AI/ML or Data Analytics

In AI/ML and Data Analytics, between() is used to select observations within a numerical range during exploratory data analysis, data cleaning, filtering, and preprocessing.

# Topic -8  Sorting Data

## Step 1 – Understand the Concept

Sorting Data means arranging the rows of a DataFrame based on the values in one or more columns.

The `sort_values()` method is used for sorting.

General syntax:

`df.sort_values("Column")`

By default, the values are sorted in ascending order.

For descending order, use:

`df.sort_values("Column", ascending=False)`

Ascending order arranges values from small to large.

Descending order arranges values from large to small.

## Step 2 – Demonstrate the Concept

### Real-World Analogy

Think of a school ranking list. Students can be arranged from the highest mark to the lowest mark. This is an example of sorting.

### Simple Example

Suppose we have the following customer table:

| Name | Age | Purchase |
|------|-----|----------|
| Arun | 20 | 3000 |
| Priya | 25 | 1000 |
| Ravi | 30 | 4000 |
| Divya | 35 | 2000 |

If we sort by Purchase in ascending order:

| Name | Age | Purchase |
|------|-----|----------|
| Priya | 25 | 1000 |
| Divya | 35 | 2000 |
| Arun | 20 | 3000 |
| Ravi | 30 | 4000 |

If we sort by Purchase in descending order, the highest purchase will appear first.

### Business Example

A company can sort customers by purchase amount to identify high-value customers, or sort products by sales to find the best-selling products.

### AI/ML Use Case

In AI/ML and Data Analytics, sorting is useful for exploring data, ranking observations, identifying highest or lowest values, and understanding patterns in a dataset.

```python
# Import Pandas
import pandas as pd

# Create customer data
customer_data = {
    "Name": ["Arun", "Priya", "Ravi", "Divya"],
    "Age": [20, 25, 30, 35],
    "Purchase": [3000, 1000, 4000, 2000]
}

# Create a DataFrame
customer_df = pd.DataFrame(customer_data)

# Sort the DataFrame by Purchase in ascending order
ascending_df = customer_df.sort_values("Purchase")

# Display the ascending order
print("Ascending order:")
print(ascending_df)

# Sort the DataFrame by Purchase in descending order
descending_df = customer_df.sort_values(
    "Purchase",
    ascending=False
)

# Display the descending order
print("\nDescending order:")
print(descending_df)
```

### Where It Is Used in AI/ML or Data Analytics

In AI/ML and Data Analytics, sorting is used during exploratory data analysis to rank records, identify extreme values, compare observations, and understand patterns in datasets.
