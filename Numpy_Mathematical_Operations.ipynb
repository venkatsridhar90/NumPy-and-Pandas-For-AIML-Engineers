# NumPy Mathematical Operations

```python
import numpy as np

# Rows: students; columns: Math, Science, English
marks = np.array([
    [80, 75, 90],
    [65, 70, 72],
    [90, 88, 95],
    [55, 60, 58],
    [75, 82, 78]
])

print("Marks dataset:")
print(marks)
print("Shape:", marks.shape)
```

## 1. Sum 
- Adds all values or values along a selected axis.
- Useful for finding total marks, sales, or quantities.

```python
print("Sum of all marks:", np.sum(marks))
```

## 2. Mean
- Calculates the arithmetic average of values.
- Useful for finding average marks, salary, or sales.

```python
print("Mean of all marks:", np.mean(marks))
```

## 3. Median 
- Finds the middle value after sorting the data.
- It is less affected by extreme values than the mean.

```python
print("Median of all marks:", np.median(marks))
```

## 4. Variance 

- Measures how far values are spread around the mean, using squared differences.
- A larger variance means the values are more spread out.

```python
print("Variance of all marks:", np.var(marks))
```

## 5. Standard Deviation 
- Measures the typical spread of values around the mean.
- It is the square root of variance and uses the same units as the data.

```python
print("Standard deviation of all marks:", np.std(marks))
```

## 6. Minimum 
- Returns the smallest value in the data.
- Useful for finding the lowest mark, price, or measurement.

```python
print("Minimum mark in the dataset:", np.min(marks))
```

## 7. Maximum 
- Returns the largest value in the data.
- Useful for finding the highest mark, price, or measurement.

```python
print("Maximum mark in the dataset:", np.max(marks))
print("Maximum for each subject:", np.max(marks, axis=0))
```

## 8. Argmin 
- Returns the index position of the smallest value in a flattened array by default.
- Use `axis` to find the index within each row or column.

```python
print("Index of the smallest value in the flattened array:", np.argmin(marks))
print("Index of minimum in each subject column:", np.argmin(marks, axis=0))
```

## 9. Argmax 
- Returns the index position of the largest value in a flattened array by default.
- Useful for locating the position of the highest value.

```python
print("Index of the largest value in the flattened array:", np.argmax(marks))
print("Index of maximum in each subject column:", np.argmax(marks, axis=0))
```

## 10. Percentile 
- Returns the value below which a given percentage of observations fall.
- For example, the 50th percentile is the median.

```python
print("50th percentile of all marks:", np.percentile(marks, 50))
print("75th percentile for each subject:", np.percentile(marks, 75, axis=0))
```

## 11. Dot Product 
- Multiplies corresponding elements of two vectors and adds the results.
- It can calculate a weighted total, such as marks multiplied by subject weights.

```python
student_marks = marks[0]       # First student's marks
subject_weights = np.array([0.4, 0.3, 0.3])

weighted_total = np.dot(student_marks, subject_weights)
print("First student's marks:", student_marks)
print("Subject weights:", subject_weights)
print("Weighted total:", weighted_total)
```

## 12. Matrix Multiplication
- Multiplies rows of the first matrix by columns of the second matrix.
- The number of columns in the first matrix must equal the number of rows in the second.

```python
# Use the first two students' marks and a 3x2 transformation matrix
two_students = marks[:2]
transform = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

result = two_students @ transform
print("First two students' marks:\\n", two_students)
print("Transformation matrix:\\n", transform)
print("Matrix multiplication result:\\n", result)
```

## 13. Transpose 
- Changes rows into columns and columns into rows.
- Useful when data needs to be arranged in a different orientation.

```python
transposed_marks = marks.T
print("Original shape:", marks.shape)
print("Transposed shape:", transposed_marks.shape)
print(transposed_marks)
```

## 14. Basic Linear Algebra Operations
- Linear algebra operations work with vectors and matrices, including solving equations and finding determinants.
- They are used in machine learning for transformations, regression, and many model calculations.

```python
# Use a small 2x2 matrix for common linear algebra operations
A = np.array([[2, 1],
              [1, 3]])
b = np.array([8, 13])

print("Matrix A:\\n", A)


print("Determinant of A:", np.linalg.det(A))


print("Inverse of A:\\n", np.linalg.inv(A))


print("Solution to A x = b:", np.linalg.solve(A, b))


print("Eigenvalues of A:", np.linalg.eigvals(A))
```
