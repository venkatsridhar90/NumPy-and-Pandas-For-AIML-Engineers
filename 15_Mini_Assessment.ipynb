## Question 7 – Filter Rows Using Multiple Conditions

Create a DataFrame containing:

- Name

Filter and display students who satisfy both conditions:

# Section 4 – Combining Data

## Question 8 – Merge Two DataFrames

Create two DataFrames:

### Customer Data
- Customer_ID
- Name

### Order Data
- Customer_ID
- Order_Amount

Merge the two DataFrames using Customer_ID.

Display the merged DataFrame.

```python
import pandas as pd

customers = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Name": ["Arun", "Priya", "Kumar"]
})

orders = pd.DataFrame({
    "Customer_ID": [101, 102, 103],
    "Order_Amount": [5000, 7000, 3000]
})

merged_data = pd.merge(
    customers,
    orders,
    on="Customer_ID",
    how="inner"
)

print(merged_data)
```

# Section 5 – Grouping and Aggregation

## Question 9 – GroupBy and Aggregation

Create a sales DataFrame containing:

- Employee
- Department
- Sales

Perform the following:

1. Group the data by Department.
2. Calculate total sales for each department.
3. Calculate average sales for each department.

Display the results.

```python
import pandas as pd

df = pd.DataFrame({
    "Employee": ["Arun", "Priya", "Kumar", "Anitha", "Rahul", "Divya"],
    "Department": [
        "Sales", "Sales", "HR",
        "HR", "Sales", "Finance"
    ],
    "Sales": [50000, 60000, 30000, 35000, 45000, 40000]
})

department_summary = df.groupby("Department")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean"
)

print(department_summary)
```

# Section 6 – Missing Values


## Question 10 – Handle Missing Values

Create a DataFrame containing missing values in at least two columns.

Perform the following:

1. Identify the missing values.
2. Count the missing values in each column.
3. Handle the missing numerical values using an appropriate method.
4. Display the cleaned DataFrame.

```python
import pandas as pd

df = pd.DataFrame({
    "Name": ["Arun", "Priya", "Kumar", "Anitha"],
    "Age": [22, None, 24, 25],
    "Marks": [85, 90, None, 88]
})

print("Missing values:")
print(df.isnull())

print("\nMissing value count:")
print(df.isnull().sum())

# Fill missing values using median
df["Age"] = df["Age"].fillna(df["Age"].median())
df["Marks"] = df["Marks"].fillna(df["Marks"].median())

print("\nCleaned DataFrame:")
print(df)
```

# Section 8 – Memory Optimization

## Question 12 – Optimize a DataFrame for Memory Usage

Create a DataFrame containing:

- Age
- Sales
- City

Perform the following:

1. Check the data types.
2. Check the memory usage.
3. Convert suitable numerical columns to smaller data types.
4. Convert the City column to the category data type.
5. Check the memory usage again.
6. Compare the memory usage before and after optimization.

```python
import pandas as pd

df = pd.DataFrame({
    "Age": [22, 24, 25, 30, 28],
    "Sales": [1000, 2000, 1500, 3000, 2500],
    "City": ["Chennai", "Mumbai", "Chennai", "Delhi", "Mumbai"]
})

print("Data Types Before Optimization:")
print(df.dtypes)

memory_before = df.memory_usage(deep=True).sum()

print("\nMemory Usage Before Optimization:")
print(memory_before, "bytes")

# Optimize data types
df["Age"] = df["Age"].astype("int8")
df["Sales"] = df["Sales"].astype("int32")
df["City"] = df["City"].astype("category")

print("\nData Types After Optimization:")
print(df.dtypes)

memory_after = df.memory_usage(deep=True).sum()

print("\nMemory Usage After Optimization:")
print(memory_after, "bytes")

print("\nMemory Saved:")
print(memory_before - memory_after, "bytes")
```

## Question 13 – Complete Data Analysis Exercise

Create a Pandas DataFrame containing at least 10 sales records with:

- Customer_ID
- Product
- Category
- City
- Quantity
- Sales

Perform the following tasks:

1. Display the first five rows.
2. Check the DataFrame shape.
3. Check the data types.
4. Check for missing values.
5. Filter records where Sales is greater than 20,000.
6. Group the data by Category.
7. Calculate total Sales for each Category.
8. Calculate average Sales for each Category.
9. Sort the results by total Sales in descending order.
10. Optimize suitable columns for memory usage.

```python
import pandas as pd

df = pd.DataFrame({
    "Customer_ID": [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    "Product": [
        "Laptop", "Phone", "Tablet", "Laptop", "Phone",
        "Monitor", "Tablet", "Laptop", "Phone", "Monitor"
    ],
    "Category": [
        "Electronics", "Electronics", "Electronics",
        "Electronics", "Electronics", "Accessories",
        "Electronics", "Electronics", "Electronics", "Accessories"
    ],
    "City": [
        "Chennai", "Mumbai", "Delhi", "Chennai", "Bangalore",
        "Chennai", "Mumbai", "Delhi", "Chennai", "Bangalore"
    ],
    "Quantity": [5, 10, 7, 4, 8, 6, 9, 3, 12, 5],
    "Sales": [50000, 30000, 20000, 45000, 28000,
              15000, 25000, 35000, 32000, 18000]
})

# 1. Display first five rows
print("First Five Rows:")
print(df.head())

# 2. Check shape
print("\nShape:")
print(df.shape)

# 3. Check data types
print("\nData Types:")
print(df.dtypes)

# 4. Check missing values
print("\nMissing Values:")
print(df.isnull().sum())

# 5. Filter Sales greater than 20,000
high_sales = df[df["Sales"] > 20000]

print("\nSales Greater Than 20,000:")
print(high_sales)

# 6 & 7 & 8. GroupBy and Aggregation
category_summary = df.groupby("Category")["Sales"].agg(
    Total_Sales="sum",
    Average_Sales="mean"
)

# 9. Sort by Total Sales
category_summary = category_summary.sort_values(
    "Total_Sales",
    ascending=False
)

print("\nCategory Summary:")
print(category_summary)

# 10. Optimize suitable columns
df["Customer_ID"] = df["Customer_ID"].astype("int16")
df["Quantity"] = df["Quantity"].astype("int8")
df["Category"] = df["Category"].astype("category")
df["City"] = df["City"].astype("category")

print("\nOptimized Data Types:")
print(df.dtypes)

print("\nOptimized Memory Usage:")
print(df.memory_usage(deep=True))
```
